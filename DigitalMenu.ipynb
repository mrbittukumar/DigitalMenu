```python
print("~~~~~~~~~~~~~~~~~~~~~~~")
print("WELCOME TO MY RESTURANT")
print("~~~~~~~~~~~~~~~~~~~~~~~")
Menu = {
    "Pizza" : 60,
    "Burger" : 50,
    "Tea" : 10,
    "Coffee" : 20,
    "Samosa" : 10
}
print("Pizza = Rs60\nBurger = Rs50\nTea = Rs10\nCoffee = Rs20\nSamosa = Rs10\n")

total = 0
while True:
    item1 = input("\nEnter The Food Items\n")

    if item1 in Menu:
        print(f"\nYour Item {item1} has been added\n")
        total += Menu[item1]
        print(f"\nYour total bill is = Rs{total} \n")
        break
    else:
       print("\nItem not available choose item from menu\n")

    #2nd order start..
while True:
    item2 = input("\nIf you want to order more item then enter the item name\n Or enter no to finish\n")

    if item2.lower() == 'no':
       break

    if item2 in Menu:
       print(f"\nYour item {item2} has been added\n")
       total += Menu[item2]
       print(f"Your total bill is = Rs{total}")
    else:
       print("item not available")

    #final...

print("==============================")
print(f"Your Total Bill is = Rs{total}")
print("==============================")

print("*************************")
print("Thankyou For Ordering 👍")
print("*************************")
```
